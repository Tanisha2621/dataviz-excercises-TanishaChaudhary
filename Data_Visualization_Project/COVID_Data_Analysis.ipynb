```python
import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go
```

```python
## Load the Dataset
df = pd.read_csv("../data/Covid_Data.csv")
```

# Dataset Overview

```python
df.shape
df.head()
df.info()
df.describe()
```

```python
# Check for missing values and duplicates
df.isnull().sum()
df.duplicated().sum()
```

### Cleaning notes

Raw fields use coded values: `1` = Yes, `2` = No, `97/98/99` = missing. `SEX`: `1` = Female, `2` = Male.
`DATE_DIED` is `9999-99-99` for patients who survived, or the death date otherwise.

```python
covid_df = df.copy()

# Create readable columns
covid_df['DIED'] = covid_df['DATE_DIED'] != '9999-99-99'
covid_df['DEATH_DATE'] = pd.to_datetime(covid_df['DATE_DIED'], format='%d/%m/%Y', errors='coerce')
covid_df['SEX_LABEL'] = covid_df['SEX'].map({1: 'Female', 2: 'Male'})
covid_df['HOSPITALIZED'] = covid_df['PATIENT_TYPE'] == 2
covid_df['COVID_POSITIVE'] = covid_df['CLASIFFICATION_FINAL'].isin([1, 2, 3])

# Age groups
covid_df['AGE_GROUP'] = pd.cut(
    covid_df['AGE'],
    bins=[0, 9, 19, 29, 39, 49, 59, 69, 79, 89, 120],
    labels=['0-9', '10-19', '20-29', '30-39', '40-49', '50-59', '60-69', '70-79', '80-89', '90+'],
    include_lowest=True
)

# Keep confirmed COVID-19 cases only
covid_df = covid_df[covid_df['COVID_POSITIVE']]
covid_df.shape
```

```python
# Count of pre-existing conditions per patient
comorbid_cols = ['DIABETES', 'COPD', 'ASTHMA', 'INMSUPR', 'HIPERTENSION',
                 'OTHER_DISEASE', 'CARDIOVASCULAR', 'OBESITY', 'RENAL_CHRONIC', 'TOBACCO']

covid_df['COMORBID_COUNT'] = (covid_df[comorbid_cols] == 1).sum(axis=1)
covid_df[comorbid_cols + ['COMORBID_COUNT']].head()
```

## Question 1 — How does mortality rate vary by age group and sex?

```python
age_sex = (covid_df.groupby(['AGE_GROUP', 'SEX_LABEL'], observed=True)['DIED'].mean() * 100).unstack()

fig = px.imshow(
    age_sex.T,
    color_continuous_scale="Blues",
    text_auto=".0f",
    title="COVID-19 Mortality Rate (%) by Age Group and Sex"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Age Group",
    yaxis_title="Sex",
    font=dict(size=14)
)

fig.show()
```

# Insight
    Mortality rises sharply from age 50 onward for both sexes. At every age group, men have a higher mortality rate than women, with the widest gap in the 50-69 age range.

## Question 2 — Does mortality rate increase with the number of pre-existing conditions?

```python
comorbid_mortality = (covid_df.groupby('COMORBID_COUNT')['DIED'].mean() * 100).reset_index()
comorbid_mortality = comorbid_mortality[comorbid_mortality['COMORBID_COUNT'] <= 6]

fig = px.bar(
    comorbid_mortality,
    x="COMORBID_COUNT",
    y="DIED",
    color="DIED",
    color_continuous_scale="Blues",
    text_auto=".0f",
    title="Mortality Rate by Number of Pre-Existing Conditions"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Number of Conditions",
    yaxis_title="Mortality Rate (%)",
    coloraxis_showscale=False,
    font=dict(size=14)
)

fig.show()
```

# Insight
    Mortality rate climbs steadily as the number of pre-existing conditions increases, from around 7% with no conditions to over 40% with four or more.

## Question 3 — Which pre-existing condition has the highest mortality rate?

```python
condition_mortality = []
for c in comorbid_cols:
    rate = covid_df[covid_df[c] == 1]['DIED'].mean() * 100
    condition_mortality.append({"Condition": c.replace("_", " ").title(), "Mortality_Rate": rate})

condition_df = pd.DataFrame(condition_mortality).sort_values("Mortality_Rate")

fig = px.bar(
    condition_df,
    x="Mortality_Rate",
    y="Condition",
    orientation="h",
    color="Mortality_Rate",
    color_continuous_scale="Blues",
    text_auto=".0f",
    title="Mortality Rate by Pre-Existing Condition"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Mortality Rate (%)",
    yaxis_title="Condition",
    coloraxis_showscale=False,
    font=dict(size=14)
)

fig.show()
```

# Insight
    Chronic kidney disease (Renal Chronic) has the highest mortality rate of any single condition, followed by COPD and cardiovascular disease.

## Question 4 — Does obesity or diabetes drive mortality more, and what happens when a patient has both?

```python
combo = covid_df[covid_df['OBESITY'].isin([1, 2]) & covid_df['DIABETES'].isin([1, 2])].copy()
combo['Obesity'] = combo['OBESITY'].map({1: 'Obese', 2: 'Not Obese'})
combo['Diabetes'] = combo['DIABETES'].map({1: 'Diabetic', 2: 'Not Diabetic'})

combo_mortality = (combo.groupby(['Obesity', 'Diabetes'])['DIED'].mean() * 100).reset_index()
combo_mortality['Group'] = combo_mortality['Obesity'] + " & " + combo_mortality['Diabetes']

fig = px.bar(
    combo_mortality,
    x="Group",
    y="DIED",
    color="DIED",
    color_continuous_scale="Blues",
    text_auto=".0f",
    title="Mortality Rate by Obesity and Diabetes Status"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="",
    yaxis_title="Mortality Rate (%)",
    coloraxis_showscale=False,
    font=dict(size=14)
)

fig.show()
```

# Insight
    Diabetes drives mortality risk much more than obesity does. Patients with diabetes alone have a similar mortality rate to patients with both conditions, showing diabetes is the dominant risk factor here.

## Question 5 — Among hospitalized patients, is ICU admission more common for men than women?

```python
hospitalized = covid_df[covid_df['HOSPITALIZED'] & covid_df['ICU'].isin([1, 2])].copy()
hospitalized['ICU_YES'] = hospitalized['ICU'] == 1

icu_by_sex = (hospitalized.groupby('SEX_LABEL')['ICU_YES'].mean() * 100).reset_index(name="ICU_Rate")

fig = px.bar(
    icu_by_sex,
    x="SEX_LABEL",
    y="ICU_Rate",
    color="ICU_Rate",
    color_continuous_scale="Blues",
    text_auto=".1f",
    title="ICU Admission Rate by Sex (Hospitalized Patients)"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Sex",
    yaxis_title="ICU Rate (%)",
    coloraxis_showscale=False,
    font=dict(size=14)
)

fig.show()
```

# Insight
    Hospitalized men are admitted to the ICU slightly more often than hospitalized women.

## Question 6 — Which medical unit has the highest ICU utilization?

```python
icu_data = covid_df[covid_df['HOSPITALIZED'] & covid_df['ICU'].isin([1, 2])].copy()
icu_data['ICU_YES'] = icu_data['ICU'] == 1

unit_counts = icu_data['MEDICAL_UNIT'].value_counts()
valid_units = unit_counts[unit_counts > 500].index
icu_data = icu_data[icu_data['MEDICAL_UNIT'].isin(valid_units)]

icu_by_unit = (icu_data.groupby('MEDICAL_UNIT')['ICU_YES'].mean() * 100).reset_index(name="ICU_Rate")
icu_by_unit['MEDICAL_UNIT'] = "Unit " + icu_by_unit['MEDICAL_UNIT'].astype(str)

fig = px.bar(
    icu_by_unit,
    x="MEDICAL_UNIT",
    y="ICU_Rate",
    color="ICU_Rate",
    color_continuous_scale="Blues",
    text_auto=".0f",
    title="ICU Admission Rate by Medical Unit"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Medical Unit",
    yaxis_title="ICU Rate (%)",
    coloraxis_showscale=False,
    font=dict(size=14)
)

fig.show()
```

# Insight
    ICU utilization varies a lot across medical unit types — some units rarely admit patients to ICU, while others do so for over a quarter of hospitalized patients.

## Question 7 — How did weekly deaths trend over time for hospitalized vs. outpatient cases?

```python
deaths = covid_df[covid_df['DIED']].copy()
deaths['Week'] = deaths['DEATH_DATE'].dt.to_period('W').dt.start_time
deaths['Status'] = deaths['HOSPITALIZED'].map({True: 'Hospitalized', False: 'Outpatient'})
deaths = deaths[deaths['Week'] <= '2020-09-01']

weekly_deaths = deaths.groupby(['Week', 'Status']).size().reset_index(name="Deaths")

fig = px.line(
    weekly_deaths,
    x="Week",
    y="Deaths",
    color="Status",
    markers=True,
    title="Weekly COVID-19 Deaths by Patient Status"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Week",
    yaxis_title="Deaths",
    font=dict(size=14)
)

fig.show()
```

# Insight
    Weekly deaths peaked in early July 2020, and hospitalized patients account for almost all deaths at every point in time — outpatient deaths stay low throughout.

## Question 8 — Do pregnant women face a higher mortality risk than non-pregnant women of the same age?

```python
females = covid_df[
    (covid_df['SEX_LABEL'] == 'Female') &
    (covid_df['AGE'].between(12, 50)) &
    (covid_df['PREGNANT'].isin([1, 2]))
].copy()
females['Group'] = females['PREGNANT'].map({1: 'Pregnant', 2: 'Not Pregnant'})

preg_mortality = (females.groupby('Group')['DIED'].mean() * 100).reset_index()

fig = px.bar(
    preg_mortality,
    x="Group",
    y="DIED",
    color="DIED",
    color_continuous_scale="Blues",
    text_auto=".1f",
    title="Mortality Rate: Pregnant vs. Non-Pregnant Women"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="",
    yaxis_title="Mortality Rate (%)",
    coloraxis_showscale=False,
    font=dict(size=14)
)

fig.show()
```

# Insight
    Pregnant women (ages 12-50) do not show a higher mortality rate than non-pregnant women of the same age range — the rates are close, with pregnant women's rate slightly lower.

## Question 9 — Does the number of pre-existing conditions differ between patients who survived and those who died?

```python
covid_df['Outcome'] = covid_df['DIED'].map({True: 'Deceased', False: 'Survived'})

fig = px.box(
    covid_df,
    x="Outcome",
    y="COMORBID_COUNT",
    color="Outcome",
    title="Pre-Existing Condition Count by Outcome"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Outcome",
    yaxis_title="Number of Conditions",
    font=dict(size=14)
)

fig.show()
```

# Insight
    Patients who died carried more pre-existing conditions on average than those who survived, confirming the link found in Question 2.

## Question 10 — Which combination of age and condition count carries the highest mortality risk?

```python
covid_df['COMORBID_BAND'] = pd.cut(
    covid_df['COMORBID_COUNT'], bins=[-1, 0, 1, 2, 10], labels=['0', '1', '2', '3+']
)

age_comorbid = (covid_df.groupby(['AGE_GROUP', 'COMORBID_BAND'], observed=True)['DIED'].mean() * 100).unstack()

fig = px.imshow(
    age_comorbid.T,
    color_continuous_scale="Blues",
    text_auto=".0f",
    title="Mortality Rate (%) by Age Group and Number of Conditions"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Age Group",
    yaxis_title="Number of Conditions",
    font=dict(size=14)
)

fig.show()
```

# Insight
    Age matters more than condition count — even a patient in their 70s with zero conditions has a higher mortality rate than a patient in their 20s with three or more conditions.
